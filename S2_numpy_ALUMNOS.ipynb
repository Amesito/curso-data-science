## 🎬 Demo: ¿por qué NumPy? (solo ejecuta y flipa)

Vamos a sumar 1 a diez millones de números, con Python puro y con NumPy:

````python
lista = list(range(10_000_000))
array = np.array(lista)

print("Python puro (bucle):")
%timeit [x + 1 for x in lista]

print("NumPy (vectorizado):")
%timeit array + 1
````

# Bloque C — Indexing y slicing (elegir trozos)

Funciona como en las listas, pero con más potencia. Ejemplo:

````python
a = np.array([10, 20, 30, 40, 50, 60])
print(a[0])      # primero → 10
print(a[-1])     # último → 60
print(a[1:4])    # del índice 1 al 3 → [20 30 40]
print(a[::2])    # de 2 en 2 → [10 30 50]

m = np.arange(1, 13).reshape(3, 4)
print(m[0, 2])   # fila 0, columna 2 → 3
print(m[1])      # fila 1 entera
print(m[:, 0])   # columna 0 entera (¡el : significa "todas las filas"!)
````

# Bloque D — Vectorización: adiós a los bucles

Aquí está la magia. Las operaciones se aplican a TODO el array de una vez:

````python
precios = np.array([10, 20, 30])
print(precios * 2)          # [20 40 60]
print(precios + 5)          # [15 25 35]
print(precios > 15)         # [False  True  True]
print(precios.sum(), precios.mean(), precios.max(), precios.min())
````

### Ejercicio 13
Calcula sobre el array `notas_clase`:
- `media` → nota media
- `desviacion` → desviación típica (pista: existe `.std()`)
- `rango` → diferencia entre la nota máxima y la mínima

````python
notas_clase = np.array([5.5, 7.0, 4.0, 9.5, 6.0, 8.0, 3.5, 7.5])
# ✏️ TU CÓDIGO
media = notas_clase.sum() / len(notas_clase)
desviacion = notas_clase.std()
rango = notas_clase.max() - notas_clase.min()
````

````python
# ✅ AUTOCORRECCIÓN
assert round(media, 2) == 6.38
assert round(desviacion, 2) == 1.90
assert rango == 6.0
print("✅ Ejercicio 13 correcto")
````

# Bloque E — Máscaras booleanas (filtrar como un pro)

El truco más importante de hoy. Una condición devuelve un array de True/False, y ese array **filtra**:

````python
edades = np.array([15, 22, 17, 34, 18, 12])
mayores = edades[edades >= 18]     # filtra: solo los que cumplen
print(mayores)                      # [22 34 18]
print((edades >= 18).sum())        # cuenta cuántos cumplen → 3
````

- `num_dias_calor` → cuántos días fueron
