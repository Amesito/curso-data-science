# 🧹 Sesión 6 — Limpieza de datos: el trabajo sucio (y más importante)

**La teoría en 4 líneas:**
- Los datos reales **siempre** llegan sucios: huecos (`NaN`), duplicados, texto mal escrito y números guardados como texto.
- Un data scientist pasa más de la mitad de su tiempo limpiando. Si limpias mal, todo lo que venga después estará mal.
- Para cada problema hay 3 caminos: **eliminar**, **rellenar** o **deducir**. Elegir bien es criterio, no código.
- Regla de oro: **trabaja siempre sobre una copia** y comprueba cuántas filas tienes antes y después de cada paso.

Hoy limpiarás dos cosas: los videojuegos (con sus nulos y duplicados reales) y el inventario de una tienda retro muy desordenada.

⚠️ **El nombre de la variable forma parte del enunciado.**

# 🏆 RETO FINAL — Automatiza la limpieza

La tienda te va a mandar inventarios nuevos **cada semana**, igual de sucios. No vas a repetir 15 ejercicios cada lunes: escribe una **función** que lo haga todo.

Escribe `limpiar_inventario(df)` que reciba un inventario sucio y devuelva uno limpio, haciendo en este orden:
1. Trabajar sobre una copia (`df = df.copy()`)
2. `Producto` → sin espacios y en formato título
3. `Consola` → sin espacios y en mayúsculas
4. `Estado` → sin espacios y en minúsculas
5. `Precio` → número decimal (`€` fuera, coma a punto, coerce)
6. `Stock` → entero, con los `"?"` como 0
7. Eliminar duplicados exactos
8. Eliminar las filas sin precio (no se puede vender algo sin precio)
9. Añadir una columna `Valor` = `Precio` × `Stock`
10. Devolver el resultado con `.reset_index(drop=True)` (para que las filas vuelvan a numerarse 0, 1, 2...)

````python
# ✏️ TU CÓDIGO
def limpiar_inventario(df):
    limpios = df.copy()
    limpios["Producto"] = limpios["Producto"].str.strip().str.title()
    limpios["Consola"] = limpios["Consola"].str.strip().str.upper()
    limpios["Estado"] = limpios["Estado"].str.strip().str.lower()
    limpios["Precio"] = pd.to_numeric(
        limpios["Precio"].astype(str)
        .str.replace("€", "", regex=False)
        .str.replace(",", ".", regex=False)
        .str.strip(),
        errors="coerce"
    )
    limpios["Stock"] = pd.to_numeric(
        limpios["Stock"].astype(str).str.replace("?", "0", regex=False),
        errors="coerce"
    ).fillna(0).astype(int)
    limpios = limpios.drop_duplicates()
    limpios = limpios.dropna(subset=["Precio"])
    limpios["Valor"] = limpios["Precio"] * limpios["Stock"]
    return limpios.reset_index(drop=True)


inventario_limpio = limpiar_inventario(inventario)
print(inventario_limpio)



````
